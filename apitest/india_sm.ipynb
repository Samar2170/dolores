```python
def is_number(value):
    if value is None:
        return False
    try:
        float(value)
        return True
    except (ValueError, TypeError):
        return False

def normalize_financial_record(record):
    statements = record["stockFinancialMap"]

    return {
        "fiscal_year": record["FiscalYear"],
        "period_end": record["EndDate"],
        "type": record["Type"],

        "income_statement": {
            item["key"]: float(item["value"])
            for item in (statements.get("INC") or [])
            if is_number(item["value"])
        },

        "balance_sheet": {
            item["key"]: float(item["value"])
            for item in (statements.get("BAL") or [])
            if is_number(item["value"])
        },

        "cash_flow": {
            item["key"]: float(item["value"])
            for item in (statements.get("CAS") or [])
            if is_number(item["value"])
        }
    }



normalized_financials = [normalize_financial_record(record) for record in data['financials']]
normalized_financials
```
